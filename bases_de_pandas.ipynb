## Lecture de données depuis une base SQL

Pour accéder à une base de données, il faut une "connexion".
- dans le cas de SQLite, on utilise le module sqlite3
- dans le cas d'autres bases, on utilise le module sqlalchemy (https://www.kdnuggets.com/using-sql-with-python-sqlalchemy-and-pandas)

### Information à demander

* Authentification : login / mot de passe, ou clé d'accès
* Paramètres de connexion : adresse réseau de la base, nom du serveur, etc.
* Code Python pour se connecter (SQL Alchemy, qui fournit une connexion)
* Eventuellement, la ou les requêtes pour récupérer les données

```python
import sqlite3

# Création d'une connexion vers la base de données
cnx = sqlite3.connect('population.sqlite')
```

```python
pd.read_sql_query('SELECT * FROM population', cnx).head(5)
```

```python
pd.read_sql('SELECT * FROM population', cnx).head()
```

```python
# Utilisation de la clause Where
# Ici : récupération des villes aux USA
pd.read_sql('SELECT * FROM population WHERE Country = "USA" ', cnx).shape
```

```python
# Utilisation d'une connexion directement dans l'instruction
pd.read_sql_query('SELECT * FROM population', sqlite3.connect('population.sqlite')).head(3)
```

### "Stratégie" de remplacement des valeurs manquantes

- diagnostiquer la présence de valeurs manquantes
- choisir la bonne stratégie pour les traiter
    * supprimer les observations avec des valeurs manquantes : `.dropna(subset=[liste des variables à tester])`
    * choisir une valeur de remplacement (moyenne, médiane, mode, valeur fixe par défaut 0)
    * utiliser des règles "métier" ou "logique" pour déterminer la valeur à utiliser

```python
cities_df.fillna({'Population': cities_df['Population'].mean(),
                  'Density': 0,
                  'Land area': cities_df['Land area'].median(),
                  'Country': cities_df['Country'].mode()
                 })
```

### L'astuce de la parenthèse (_parenthesis trick_)

```python
(cities_df
 # Regroupement par pays
 .groupby(['Country'])
 
 # Calcul des agrégats
 .agg({'Population': ['mean', 'max', 'sum'], 'Land area': ['mean', 'min', 'max']})
 
 .head(3)
)
```

```python
(cities_df
 # Regroupement par pays
 .groupby(['Country'])
 
 # Calcul des agrégats
 .agg(pop_total=('Population', 'sum'), nb_cities=('City / Urban area', 'count'))
 
 # Tri par nombre de villes décroissant
 .sort_values(['nb_cities'], ascending=[False])
 
 .head(3)
)
```

## Visualisation

```python
(cities_df
 .groupby(['Country'])
 .agg(pop_total=('Population', 'sum'), nb_cities=('City / Urban area', 'count'))
 .sort_values(['nb_cities'], ascending=[False])
).plot(kind='bar', y='nb_cities', figsize=(12, 6))
```

```python
import seaborn as sns

sns.set()

(cities_df
 .groupby(['Country'])
 .agg(pop_total=('Population', 'sum'), nb_cities=('City / Urban area', 'count'))
 .sort_values(['nb_cities'], ascending=[False])
).plot(kind='bar', y='nb_cities', figsize=(12, 6))
```
