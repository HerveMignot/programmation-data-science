### Statistique de base

* dénombrement : .value_counts()
* croisement : pd.crosstab()
* statistiques : .describe()

```python
sales_df['ProductCategory'].value_counts()
```

```python
pd.crosstab(sales_df['ProductCategory'], sales_df['PaymentMethod'], normalize='all') # normalize : index, columns
```

```python
pd.crosstab(sales_df['ProductCategory'], sales_df['PaymentMethod'], normalize='columns') 
```

```python
pd.crosstab(sales_df['ProductCategory'], sales_df['PaymentMethod'],
            values=sales_df['Quantity'], aggfunc='mean') # 'sum', 'min', 'max'
```

```python
sales_df.drop(columns=['CustomerID', 'ProductID', 'StoreLocation']).describe(include='all')
```

### Représentation graphique

```python
sales_df.groupby('ProductCategory')['CustomerID'].count().plot(kind='pie', legend=True, figsize=(10, 8));
```

```python
# accessor dt pour traiter les dates
sales_df['TransactionDate'].dt.weekday
```

```python
# Calcul des ventes par jour de la semaine
sales_df.groupby(sales_df['TransactionDate'].dt.weekday)['TotalAmount'].sum()
```

```python
sales_df.groupby(sales_df['TransactionDate'].dt.weekday)['TotalAmount'].sum().plot.bar();
```

```python
import seaborn as sns

sns.set()

sales_df.groupby(sales_df['TransactionDate'].dt.weekday)['TotalAmount'].sum().plot.bar();
```

```python
sales_df['TransactionDate'].dt.date
```

```python
sales_df.groupby(sales_df['TransactionDate'].dt.date)['TotalAmount'].sum()
```
